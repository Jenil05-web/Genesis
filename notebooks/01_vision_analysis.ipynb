# Vision analysis 

In this notebook we are basically testing whether image analysisi is working properly or not

```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()
```

```python
client = OpenAI()
SAMPLE_IMAGE_URL = "https://images.unsplash.com/photo-1682351888673-9f898b62a1c9?q=80&w=1171&auto=format&fit=crop&ixlib=rb-4.1.0&ixid=M3wxMjA3fDB8MHxwaG90by1wYWdlfHx8fGVufDB8fHx8fA%3D%3D"
```

```python
# we are still just experimenting whether image api and everything is properly being detected or not

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "Describe what you see in this image, focusing on any disaster-related damage."},
                {"type": "image_url", "image_url": {"url": SAMPLE_IMAGE_URL}},
            ],
        }
    ],
)
print(response.choices[0].message.content)
```

```python
import json

STRUCTURED_PROMPT = """
Analyze this image for disaster response purposes. Return ONLY valid JSON matching exactly this schema:

{
  "flooded_zones": boolean,
  "blocked_roads": boolean,
  "collapsed_structures": boolean,
  "severity_estimate": "low" | "medium" | "high" | "critical",
  "notes": "one sentence summary"
}
"""

response = client.chat.completions.create(
    model="gpt-4o",
    response_format={"type": "json_object"},
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": STRUCTURED_PROMPT},
                {"type": "image_url", "image_url": {"url": SAMPLE_IMAGE_URL}},
            ],
        }
    ],
)

raw = response.choices[0].message.content
parsed = json.loads(raw)
parsed
```
